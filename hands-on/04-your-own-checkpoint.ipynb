## 1. Set up

We connect to the server, make sure the two helper models are installed, and make a scratch
folder, `work-04`, for the files we create. `MODELS` is the registry: the folder the server reads
its models from. The commands below install into this folder, so it must be the folder your
server serves.

````python
import importlib.util
import json
import os
import re
import shutil
import subprocess
import sys
import time
from pathlib import Path

import handson as ho

client = ho.connect()
for name in ["rf-detr-nano", "efficientnet-b0"]:
    ho.ensure_model(client, name)

MODELS = Path(os.environ.get("VISIONSERVE_MODELS") or Path.home() / ".visionserve" / "models")
WORK = Path("work-04")
WORK.mkdir(exist_ok=True)
PHOTOS = ho.photo("cat").parent          # hands-on/images: 8 sample photos
CHECKPOINT = Path(os.environ.get("CHECKPOINT") or Path.home() / ".roboflow" / "models" / "rf-detr-nano.pth")


def short(p):
    """A path with your home folder written as ~ (shorter to print)."""
    return str(p).replace(str(Path.home()), "~")


print("registry  :", short(MODELS))
if CHECKPOINT.is_file():
    print("checkpoint:", short(CHECKPOINT), f"({CHECKPOINT.stat().st_size / 1e6:.0f} MB)")
else:
    print("checkpoint:", short(CHECKPOINT), "NOT FOUND - set CHECKPOINT, or run: "
          "python -c 'from rfdetr import RFDETRNano; RFDETRNano()'")
````

**What you see**

- One line for the server, one line per helper model (`installed`).
- `registry`: the models folder. `checkpoint`: the training file we will convert, about 366 MB.

<details><summary><b>In detail</b></summary>

- `MODELS` comes from the environment variable `VISIONSERVE_MODELS`, else it is
  `~/.visionserve/models`. With the Docker server, it is the folder you mounted, for example
  `~/.visionserve_models`.
- A server only sees **its own** registry (`visionserve serve --models DIR`). If you install into
  another folder, the server does not find the model.

More: [Configuration](https://mtbui2010.github.io/visionserve/reference/configuration/).

</details>

## 2. Words you need: checkpoint, ONNX, manifest, registry

- **Checkpoint**: the file your training writes. It holds the model's **weights** (the numbers the
  model learned) and often more: the optimizer state (needed only to continue training) and the
  training settings. Only the training framework (here PyTorch) can read it.
- **ONNX**: a standard file format for a trained model. It stores the weights **and** the list of
  math steps (the *graph*). *ONNX Runtime*, the engine inside VisionServe, runs ONNX files. It does
  not need PyTorch.
- **Converting** (or *exporting*): turning the checkpoint into an ONNX file.
- **Manifest** (`manifest.yaml`): a small text file next to the ONNX file. It says how to use the
  model: the photo size, how to prepare the pixels, how to read the output, the class names and
  the licence.
- **Registry**: the folder of models. Each model is one sub-folder: an ONNX file plus its manifest.

Let us open the checkpoint and look inside. (This cell needs PyTorch. Without it, it only says so.)

````python
if importlib.util.find_spec("torch") is None or not CHECKPOINT.is_file():
    print("skipped: needs PyTorch and the checkpoint")
else:
    import torch
    ckpt = torch.load(CHECKPOINT, map_location="cpu", weights_only=False)
    print("parts of the checkpoint:", list(ckpt))
    weights = ckpt["model"]
    n = sum(t.numel() for t in weights.values())
    print(f"weights: {len(weights)} tensors, {n / 1e6:.1f} million numbers")
    for k in list(weights)[:3]:
        print(f"  {k}: shape {tuple(weights[k].shape)}")
    print("class names in the training settings:", getattr(ckpt.get("args"), "class_names", None))
    del ckpt, weights
````

**What you see**

- The checkpoint is a Python dictionary with five parts: `model` (the weights), `optimizer`,
  `lr_scheduler`, `epoch` and `args` (the training settings).
- The weights are 465 *tensors* (a tensor is an array of numbers with a shape), about 30 million
  numbers in total. One number is 4 bytes, so the weights are about 120 MB. The rest of the
  366 MB file is the optimizer state, which serving does not need.
- `class names ... None`: this official checkpoint does **not** store the class names. A checkpoint
  from your own RF-DETR training usually does. Remember this for section 6.

<details><summary><b>In detail</b></summary>

- A *tensor shape* such as `(768, 256)` means a table of 768 rows and 256 columns of numbers.
- `weights_only=False` lets PyTorch read the `args` part (a Python object). Do this only with files
  you trust: loading a checkpoint can run code.
- ONNX in more detail, and how VisionServe uses ONNX Runtime:
  [Model files and ONNX Runtime](https://mtbui2010.github.io/visionserve/concepts/onnx/).

</details>

## 3. Which command for my file?

Find your file in the first column. The first two rows read everything they need from the file.
The other formats cannot know what the model does, so you tell them with `--task` (what the model
does), `--input` (the photo size it was trained on) and `--license`.

````python
ho.table([
    ["RF-DETR training checkpoint (.pth)", "visionserve convert rfdetr best.pth --name NAME"],
    ["Hugging Face model (folder or hub id)", "visionserve convert hf google/vit-base-patch16-224 --name NAME"],
    ["TorchScript (.pt from torch.jit.save)", "visionserve convert torchscript model.pt --name NAME --task classification --input 224x224 --license MIT"],
    ["PyTorch state_dict + the Python file that builds the model", "visionserve convert pytorch weights.pth --script build.py --name NAME --task detection --input 640x640 --license MIT"],
    ["Keras (.keras, .h5)", "visionserve convert keras model.keras --name NAME --task classification --input 224x224 --license Apache-2.0"],
    ["TensorFlow Lite (float .tflite)", "visionserve convert tflite model.tflite --name NAME --task classification --input 224x224 --license Apache-2.0"],
    ["TensorFlow SavedModel folder", "visionserve convert tensorflow saved_model/ --name NAME --task classification --input 224x224 --license Apache-2.0"],
    ["An .onnx file you already have", "visionserve import model.onnx --name NAME --task detection --license Apache-2.0"],
], ["Your file", "Command"])
````

**What you see**

A table with one row per kind of file. In this notebook we use three rows: RF-DETR (section 4),
`.onnx` (section 8) and Hugging Face (section 9).

<details><summary><b>In detail</b></summary>

- `--task` can be `classification` (one score per class), `detection` (boxes from a DETR-style
  model such as RF-DETR or RT-DETR), `depth` (a depth map) or `embed` (one vector per photo).
- `convert` needs Python and PyTorch (or TensorFlow), so it runs in the **converter**: a Docker
  image, or the Python package. `import` needs neither: it is part of the Go program.
- The full table and every flag: [Use a model you trained](https://mtbui2010.github.io/visionserve/guides/use-your-model/).

</details>

## 4. Convert an RF-DETR checkpoint

First we choose **where the converter runs**. Both ways take the same flags and print the same
report:

1. **Docker**: `visionserve convert ...` starts the converter image. You need only Docker.
2. **The Python package**: `python -m visionserve.convert ...` runs the same code in your own
   Python, after `pip install "visionserve[convert]"`.

The small function below picks the Python package when PyTorch is installed in this notebook's
Python, else Docker. Set `VS_CONVERT_WITH=docker` or `VS_CONVERT_WITH=python` to choose yourself.
It also hides PyTorch's long export warnings, and prints how long the step took.

````python
CONVERT_WITH = os.environ.get("VS_CONVERT_WITH") or (
    "python" if all(importlib.util.find_spec(m) for m in ("torch", "onnx", "visionserve.convert")) else "docker")
print("the converter runs with:", CONVERT_WITH)


def _quiet(text):
    """Drop PyTorch/transformers warnings and progress bars from the converter output."""
    keep, skip_next = [], False
    for line in text.splitlines():
        if skip_next and line.startswith("  "):
            skip_next = False
            continue
        skip_next = False
        if "Warning:" in line or "it/s]" in line or line.startswith(("[transformers]", "[20")):
            skip_next = "Warning:" in line
            continue
        keep.append(line)
    return "\n".join(keep)


def convert(*args):
    """Run `visionserve convert ARGS` (Docker) or the same command from the Python package."""
    # --server: the running server that serves MODELS, for the checks B1/B2 (else a temporary one)
    args = [str(a) for a in args] + ["--models", str(MODELS), "--server", ho.host_url()]
    start = time.time()
    if CONVERT_WITH == "docker":
        out = ho.run_cli("convert", *args)
    else:
        print("$ python -m visionserve.convert " + " ".join(args))
        p = subprocess.run([sys.executable, "-m", "visionserve.convert", *args],
                           stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
        out = _quiet(p.stdout)
        print(out + f"\n(exit code {p.returncode})")
        if p.returncode not in (0, 1):     # 2 = refused or could not run
            print(f"[took {time.time() - start:.0f} s]")
            return out
    print(f"[took {time.time() - start:.0f} s]")
    return out
````

**What you see**

One line: `python` or `docker`. Nothing is converted yet.

Now the real step. We convert the checkpoint and install it under the name `my-detector`.
`--images` gives the checks real photos (our 8 sample photos). `--force` replaces an older
`my-detector`, so you can run this notebook again.

````python
out = convert("rfdetr", CHECKPOINT, "--name", "my-detector", "--images", PHOTOS, "--force")
````

**What you see** (read it from the bottom up: the verdict is at the end)

- `overall: PASS`: the model is converted, installed and checked.
- The first `rfdetr:` lines say what `convert` **read from the checkpoint**: the variant (`nano`,
  found from the shapes of the weights), the input size (384 × 384) and 91 class outputs. There
  is also a **warning**: the checkpoint has no class names, so the classes will be called
  `class_0` ... `class_90`. We fix this in section 6.
- `--- my-detector/manifest.yaml ---` shows the manifest it wrote (section 10 explains it).
- `installed my-detector ...`: the model folder is now in the registry.
- `server: ...` lines: the checks B1 and B2 need a VisionServe server that serves this registry.
  The converter uses the server we pass with `--server` (your server) when that server lists the
  new model; otherwise it starts a **temporary** server for the check and stops it after. Both are
  fine. (Here it says the server "does not list my-detector": the server updates its list at most
  once per second, and the check came less than a second after the install.)
- The table is what it checked. Each row compares with the **original model in PyTorch**:

| Row | In plain words | Here |
|---|---|---|
| **A** | The ONNX file computes the same numbers as the PyTorch model, on the same input. | largest difference about 0.001 of the value |
| **B1** | The server prepares a photo (resize, colour scaling) like the training code. | about 0.4 gray levels apart, out of 255; below 2 is a pass |
| **B2** | The server finds the same objects as the original model. | 42 of 43 boxes match, within 0.1 pixel on average |

A *gray level* is one step on the 0 to 255 scale of a pixel. 0.4 gray levels is rounding.

<details><summary><b>In detail</b></summary>

- **Why three checks?** Each one catches a different mistake. A wrong export breaks A. A wrong
  resize or colour scaling in the manifest breaks B1 (the model then sees other pictures than in
  training, and gets quietly worse). B2 checks the whole chain, decoding of the boxes included.
- **Row C** appears only with `--eval val.json` (your labelled photos): it measures the accuracy
  (mAP) of the original model and of the served one, side by side.
- "100 low-score queries differ" in row A: RF-DETR proposes 300 candidate boxes (*queries*) and
  keeps the best ones. Candidates with almost the same low score can swap places after tiny
  rounding differences. They are below the score threshold, so they are not judged one by one.
- `notes: [B2] ... threshold-boundary flips`: a box with a score very close to 0.5 can be just
  above the threshold in one model and just below in the other. Such boxes are not counted.
- A FAIL uninstalls the model again (add `--keep-on-fail` to keep it while you look).
  `--dry-run` converts and checks but installs nothing.
- `convert-report.json` in the model folder keeps the full report.
- How each check works, with real failures it caught:
  [Inspect and verify a model](https://mtbui2010.github.io/visionserve/guides/inspect/#3-let-the-converter-check-it-tiers-a-b1-b2-c).
- The converter image and how a model is installed:
  [Clients and the converter](https://mtbui2010.github.io/visionserve/architecture/clients/).

</details>

## 5. Run it

`visionserve run` loads the model in the program itself (no server needed), runs one photo and
prints the answer as JSON. It is slow because it loads the model every time, but it is the
simplest test.

````python
out = ho.run_cli("run", "my-detector", ho.photo("cat"), "--models", MODELS)
````

**What you see**

- A JSON answer: `task`, `model`, `device` (where it ran: `gpu:0` or `cpu`) and a list of
  `detections`. Each detection has a `bbox`, a `class` and a `conf`.
- `bbox` is `[x, y, width, height]` in the pixels of **your** photo (not of the 384 × 384 model
  input): VisionServe maps the boxes back for you.
- `conf` is the *confidence*: how sure the model is, from 0 to 1. Only detections above the
  manifest's `conf_threshold` (0.5) are returned.
- The classes are called `class_73`, `class_17`, `class_62`, `class_1`: **numbers, not names**.
  The model knows *that* it sees class 17, but nobody told VisionServe that class 17 is "cat".
  Next section.

<details><summary><b>In detail</b></summary>

- The last line (`predict: ... client=...ms server=...ms`) shows the time. Most of it is loading
  the model. A running server loads it once and keeps it in memory.
- `--save out.jpg` draws the boxes on the photo.
- More about the result format: [What the models do](https://mtbui2010.github.io/visionserve/concepts/tasks/).

</details>

## 6. Class names: `--labels`

A detector outputs a **class number** for each box. The *labels file* turns numbers into names:
a plain text file with one name per line, in the order of the model's outputs. Line 1 is class 0,
line 2 is class 1, and so on.

This checkpoint was trained on COCO, a public data set with 80 kinds of objects (stored as 91
numbers, some unused). The model `rf-detr-nano` that you installed already has this file:
`coco91.txt`. We copy it and convert again with `--labels`.

````python
labels = WORK / "classes.txt"
shutil.copyfile(MODELS / "rf-detr-nano" / "coco91.txt", labels)
names = labels.read_text().splitlines()
print(len(names), "names; the first five:", names[:5], "; line 18 (class 17):", names[17])

out = convert("rfdetr", CHECKPOINT, "--name", "my-detector", "--images", PHOTOS,
              "--labels", labels, "--force")
````

**What you see**

- 91 names. `N/A` marks the numbers COCO does not use. Class 17 is `cat`.
- The `rfdetr:` line now says `91 logits -> labels ['N/A', 'person', 'bicycle'] ...`: the names are
  matched to the outputs. There is no warning about class names any more.
- `server: ... already has my-detector registered ...` and `--force: previous my-detector moved to
  ...`: the converter keeps the old version aside until the new one passes its checks, and checks
  the new one on a temporary server (the running server may still hold the old one in memory).
- `overall: PASS` again, and `[3 files]` (the labels file was added to the model folder).

<details><summary><b>In detail</b></summary>

- *Logits* are the raw class scores of the model, before they become probabilities.
- The number of names must match the number of class outputs. If your file has a different
  count, `convert` tells you.
- A checkpoint from your own RF-DETR training usually stores the class names. Then you do not need
  `--labels`, and the first `rfdetr:` line shows them.

</details>

Now we use the model **through the server**, from Python, exactly like a built-in model. A
running server finds a **new** model by itself. A **replaced** model (we used `--force`) may still be
in the server's memory from an earlier run, so we unload it first: the next request then reads the
new files.

````python
try:
    client.unload("my-detector")      # only needed after replacing a model the server had loaded
except Exception:
    pass                              # it was not loaded: nothing to do

res = client.predict("my-detector", ho.photo("cat"))
for d in res.detections:
    print(f"{d.cls:10s} conf {d.conf:.2f}  bbox {[round(v) for v in d.bbox]}")
print(f"device {res.device}, {res.duration_ms:.0f} ms on the server")
ho.show(ho.photo("cat"), res)
````

**What you see**

- The same four boxes as the `run` command, now with names: `laptop`, `cat`, `chair` and
  `person` (the man; the model is less sure about him: 0.54, just above the 0.5 threshold).
- `device`: `gpu:0` when the server runs on a GPU, else `cpu`. The time is the server's time for
  this one request (the first request also loads the model, so it is slower).
- The photo with the boxes drawn on it.

<details><summary><b>In detail</b></summary>

- `client.predict(model, image)` sends the photo with an HTTP request to `/api/predict`. The Python
  client may shrink a large photo before it sends it, and maps the boxes back to your photo's
  pixels. See [Python client](https://mtbui2010.github.io/visionserve/clients/python/).
- `client.unload(name)` frees the model's memory on the server. The next request loads it again.

</details>

## 7. The licence rule: `--license`

Every model in VisionServe must declare the **licence** of the original model, and only
*permissive* licences are accepted: Apache-2.0, MIT and BSD. A permissive licence lets anyone use
the model, also in commercial and closed products.

**AGPL** models (for example Ultralytics YOLO, FastSAM, YOLO-World) are always refused. AGPL is a
*copyleft* licence: everything that is built with it must also be published under AGPL. One AGPL
model would put VisionServe, and every product that uses it, under AGPL.

The RF-DETR checkpoint did not need `--license`: `convert` knows RF-DETR is Apache-2.0. Let us see
what happens when we declare AGPL. We use `import` (next section) because it is fast; `convert`
refuses in the same way, before it exports anything.

````python
onnx_file = WORK / "exported.onnx"
shutil.copyfile(MODELS / "efficientnet-b0" / "model.onnx", onnx_file)   # an ONNX file to play with

out = ho.run_cli("import", onnx_file, "--name", "agpl-test", "--task", "classification",
                 "--license", "AGPL-3.0", "--models", MODELS)
````

**What you see**

- `FAIL: license "AGPL-3.0" is not allowed — only permissive licenses accepted`.
- `Installed: nothing installed`, and the exit code is `1`. Nothing to fix on your side: VisionServe
  cannot serve this model. Pick an Apache-2.0, MIT or BSD model instead.

<details><summary><b>In detail</b></summary>

- "It is on Hugging Face" says nothing about the licence. Read the model card or the LICENSE file
  of the **original** model, and pass that with `--license`.
- The licence is written in `manifest.yaml`, and the server checks it again every time it reads
  the registry. A manifest with a forbidden licence is not loaded.
- Why this rule exists: [Engineering rules](https://github.com/mtbui2010/visionserve/blob/main/docs/engineering-rules.md)
  and [Manifest format](https://mtbui2010.github.io/visionserve/reference/manifest/).

</details>

## 8. Already have an `.onnx` file? `visionserve import`

Maybe someone already exported the model to ONNX. Then you do not need the converter:
`visionserve import` reads the ONNX file, writes a manifest for it and installs it. It needs no
Python and no Docker.

But an ONNX file does **not** say everything. It says the input size and the output shapes. It
does not say how the photos were resized or how the colours were scaled in training. `import`
then **assumes** the most common choice, and prints each assumption so you can check it.

Our ONNX file is a copy of the `efficientnet-b0` classifier (it sorts a photo into one of 1000
ImageNet classes). We import it as `my-import`.

````python
shutil.copyfile(MODELS / "efficientnet-b0" / "imagenet1k.txt", WORK / "imagenet-classes.txt")
out = ho.run_cli("import", onnx_file, "--name", "my-import", "--task", "classification",
                 "--license", "Apache-2.0", "--labels", WORK / "imagenet-classes.txt",
                 "--force", "--models", MODELS)
````

**What you see** (a long report; look at three parts)

- `WARN: imported my-import ..., but 3 settings were assumed`. `WARN` is normal for `import`: it
  installed the model and asks you to check its guesses.
- **What import read and assumed**: `read:` lines come from the file (layout, 224 × 224 input,
  1000 class scores, the labels). `assumed:` lines are guesses:
  - **resize**: *squash* the photo to 224 × 224 (stretch it, the shape of the photo is not kept);
  - **normalisation**: the usual ImageNet colour scaling;
  - **scores**: softmax over the outputs, top 5 classes returned.
- **manifest.yaml**: the manifest it wrote. Lines marked `# assumed: check it` are the guesses.
- The rest describes the ONNX file (inputs, outputs, size), the preprocessing and the runtime.

**Is the guess right?** Here we can compare with the shipped `efficientnet-b0` manifest, written by
people who checked it:

````python
for line in (MODELS / "efficientnet-b0" / "manifest.yaml").read_text().splitlines():
    if line.strip().startswith(("preprocess:", "resize:", "crop_pct:", "resample:")) or "top-1" in line:
        print(line)
````

**What you see**

The shipped manifest says `resize: center_crop` with `crop_pct: 0.875`: shrink the photo, then cut
out the middle square. That is how these weights were trained. Its comments say that squash cost
1.6 points of accuracy (*top-1*, the share of photos whose best class is right: 77.44 % instead of
79.04 %, measured on 5000 ImageNet photos). So **the import guess (squash) is
wrong for this model**. We import again, with the right setting:

````python
out = ho.run_cli("import", onnx_file, "--name", "my-import", "--task", "classification",
                 "--license", "Apache-2.0", "--labels", WORK / "imagenet-classes.txt",
                 "--resize", "center_crop", "--force", "--models", MODELS)
print("...")
print([l for l in out.splitlines() if "settings were assumed" in l][:1])
````

**What you see**

Now `2 settings were assumed` (the resize is no longer a guess), and the manifest says
`resize: center_crop` without `# assumed`. The model is installed. Let us ask it about the cat:

````python
try:
    client.unload("my-import")       # in case an earlier run left the old version loaded
except Exception:
    pass
res = client.predict("my-import", ho.photo("cat"))
for c in res.classifications:
    print(f"{c.conf:.2f}  {c.cls}")
ho.show(ho.photo("cat"), res)
````

**What you see**

The top 5 ImageNet classes with their probability, also as a bar chart on the right. The first
answer is `laptop` (0.54), and the cat comes only as cat breeds (`tiger cat`, `tabby`,
`Egyptian cat`: ImageNet has no plain "cat" class). Why? A **classifier** gives one answer for the
**whole** photo, and this photo shows a cat **and** a large laptop. A **detector** (like
`my-detector`) gives one answer per object. Choose the task that fits your question.

**A refusal: the wrong task.** `import` checks that the file fits the task. This file is a
classifier. What if we say it is a detector?

````python
out = ho.run_cli("import", onnx_file, "--name", "my-import-det", "--task", "detection",
                 "--license", "Apache-2.0", "--models", MODELS)
````

**What you see**

`FAIL: the outputs (648 [1,1000]) are not a DETR-style set of queries`. A detector must output boxes
(`[1, Q, 4]`) and class scores (`[1, Q, classes]`). This file outputs one row of 1000 scores, so
`import` refuses and installs nothing. Fix `--task`, or use `convert` for a model `import` cannot
decode.

<details><summary><b>In detail</b></summary>

- `import` can decode three tasks from the file alone: `classification`, DETR-style `detection`
  (RF-DETR, RT-DETR) and `depth`. A YOLO export or another output layout needs `convert`.
- Flags to fix a wrong guess: `--resize squash|letterbox|center_crop|keep_aspect`,
  `--mean a,b,c`, `--std a,b,c`, `--layout NCHW|NHWC`, `--input WxH`. Use `--force` to replace the
  model.
- `WARN` means "installed, but check the assumptions". The only way to **measure** whether a guess
  is right is `visionserve check` against your labelled photos (notebook 05).
- More: [Use a model you trained, Import](https://mtbui2010.github.io/visionserve/guides/use-your-model/#already-have-an-onnx-file-import-it)
  and [From pixels to tensors](https://mtbui2010.github.io/visionserve/concepts/preprocessing/).

</details>

## 9. A Hugging Face model

[Hugging Face](https://huggingface.co) is a website where people share trained models. `convert hf`
takes a model id (or a downloaded folder) and reads the task, the input size, the colour scaling,
the class names and the licence from the model's files.

We convert `google/vit-base-patch16-224`, an image classifier (Apache-2.0). It is about 350 MB.
Set `RUN_HF = False` to skip this step.

````python
RUN_HF = True

hf_cache = Path(os.environ.get("HF_HOME") or Path.home() / ".cache" / "huggingface") / "hub"
cached = (hf_cache / "models--google--vit-base-patch16-224").is_dir()
print("already downloaded:", cached)
if RUN_HF:
    out = convert("hf", "google/vit-base-patch16-224", "--name", "my-vit", "--images", PHOTOS, "--force")
````

**What you see**

- `hf: ... model_type='vit' ... -> classification  license=Apache-2.0`: the task and the licence come
  from the model's files. No `--task`, `--input` or `--license` needed.
- `parity ... max|Δ|/scale` is check A (tiny: about 1e-6).
- `overall: WARN`, because row **B2** says `top-1 agreement 87.5% (8 images)`: on 1 of the 8
  photos, the served model's best class differs from the original's. This happens when two
  classes have almost the same score on that photo (for example 0.204 against 0.196): a small
  rounding difference swaps them. It is usually harmless. To be sure, measure accuracy with
  labelled photos (`--eval`, or `visionserve check`, notebook 05).

<details><summary><b>In detail</b></summary>

- *Top-1 agreement*: in how many photos the class with the highest score is the same for both.
- The converter supports these Hugging Face model families: GroundingDINO, SigLIP, CLIP, RT-DETR,
  image classification and depth estimation.
- With Docker, the download is kept in `~/.cache/visionserve-convert`, so it happens once.
- More: [Use a model you trained](https://mtbui2010.github.io/visionserve/guides/use-your-model/).

</details>

## 10. Where the model lives: `manifest.yaml`

Each installed model is one folder in the registry. Let us list the `my-detector` folder and read
its manifest.

````python
folder = MODELS / "my-detector"
for f in sorted(folder.iterdir()):
    print(f"{f.name:22s} {f.stat().st_size / 1e6:8.2f} MB")
print()
print((folder / "manifest.yaml").read_text())
````

**What you see**

Four files: `model.onnx` (the weights and the graph, about 108 MB: no optimizer state any more),
`manifest.yaml`, the labels file and `convert-report.json` (the full check report).

The manifest, line by line:

| Line | Meaning |
|---|---|
| `# ...` (top) | Comments: where the model came from and the results of the checks. The server ignores them. |
| `name: my-detector` | The name you use in `run`, `predict` and the API. It is also the folder name. |
| `task: detection` | What the model does: boxes with classes. |
| `license: Apache-2.0` | The licence of the original model. Only Apache-2.0, MIT and BSD load. |
| `architecture: rf-detr` | Which VisionServe code prepares the photo and reads the output. |
| `sha256: ...` | A fingerprint of `model.onnx`. If the file changes, the server refuses it. This ties the licence to these exact bytes. |
| `model_file: model.onnx` | The ONNX file in this folder. |
| `input: width / height` | The model's photo size: 384 × 384. |
| `layout: NCHW` | The order of the numbers in the input: colour channels first, then rows, then columns. |
| `letterbox: false` | *Squash* the photo to the square (stretch it). `true` would keep the shape and add grey bars. It must match training. |
| `normalize: mean / std` | Colour scaling: pixels go from 0..255 to 0..1, then minus `mean`, divided by `std`, per colour. |
| `postprocess: type: detr` | How to read the output: a DETR-style set of boxes, no NMS needed. |
| `box_format: cxcywh` | The model gives boxes as centre x, centre y, width, height. |
| `conf_threshold: 0.5` | Boxes below this confidence are dropped. |
| `max_detections: 300` | At most this many boxes. |
| `labels: ...` | The class-name file. |
| `runtime: prefer: [cuda, cpu]` | Run on the NVIDIA GPU when possible, else the CPU. |
| `idle_unload_seconds: 300` | Free the memory after 5 minutes without requests. |

<details><summary><b>In detail</b></summary>

- *NMS* (non-maximum suppression) is a step that removes duplicate boxes. YOLO models need it.
  RF-DETR does not: each of its candidate boxes already stands for a different object.
- You can edit the manifest by hand (for example `conf_threshold`). A running server reads the
  change after the model is unloaded. Do not change `sha256` or the licence by hand.
- To move the model to another machine, copy the whole folder into that machine's registry.
- Every field: [Manifest format](https://mtbui2010.github.io/visionserve/reference/manifest/). How the server finds and checks models:
  [Catalog and installing models](https://mtbui2010.github.io/visionserve/architecture/catalog/).

</details>

## Recap

- A **checkpoint** is your training file (PyTorch). **ONNX** is the portable file that VisionServe
  runs. The **manifest** says how to use it. The **registry** is the folder of models.
- `visionserve convert rfdetr best.pth --name NAME` reads the checkpoint, exports ONNX, writes the
  manifest, installs the model and **checks** it: A (same numbers), B1 (same photo preparation),
  B2 (same objects). Read `overall:` first.
- No class names in the checkpoint? Add `--labels classes.txt` (one name per line, in output order).
- Only Apache-2.0, MIT and BSD models are accepted. AGPL is always refused.
- `visionserve import model.onnx` needs no Python. Check every `assumed:` line: here the resize
  guess was wrong, and `--resize center_crop` fixed it.
- `convert hf <id>` reads task, size and licence from a Hugging Face model.
- A running server finds new models by itself. After replacing one (`--force`), unload it.
